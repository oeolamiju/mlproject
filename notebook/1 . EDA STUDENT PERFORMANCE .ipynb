### 3.8 Adding columns for "Total Score" and "Average"

```python
df['total score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average'] = df['total score']/3
df.head()
```

```python
reading_full = df[df['reading_score'] == 100]['average'].count()
writing_full = df[df['writing_score'] == 100]['average'].count()
math_full = df[df['math_score'] == 100]['average'].count()

print(f'Number of students with full marks in Maths: {math_full}')
print(f'Number of students with full marks in Writing: {writing_full}')
print(f'Number of students with full marks in Reading: {reading_full}')
```

```python
reading_less_20 = df[df['reading_score'] <= 20]['average'].count()
writing_less_20 = df[df['writing_score'] <= 20]['average'].count()
math_less_20 = df[df['math_score'] <= 20]['average'].count()

print(f'Number of students with less than 20 marks in Maths: {math_less_20}')
print(f'Number of students with less than 20 marks in Writing: {writing_less_20}')
print(f'Number of students with less than 20 marks in Reading: {reading_less_20}')
```

#####  Insights
 - From above values we get students have performed the worst in Maths 
 - Best performance is in reading section

#### BIVARIATE ANALYSIS ( Is gender has any impact on student's performance ? ) 

```python
# Calculate average scores grouped by gender

gender_group = df.groupby('gender')[['average', 'math_score']].mean()

gender_group
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))

# X-axis labels
X = ['Total Average', 'Math Average']

# Get average scores for each gender
female_scores = [
    gender_group.loc['female', 'average'],
    gender_group.loc['female', 'math_score']
]

male_scores = [
    gender_group.loc['male', 'average'],
    gender_group.loc['male', 'math_score']
]

# Create X-axis positions
X_axis = np.arange(len(X))

# Plot grouped bars
plt.bar(X_axis - 0.2, male_scores, 0.4, label='Male')

plt.bar(X_axis + 0.2, female_scores, 0.4, label='Female')

# Format chart
plt.xticks(X_axis, X)

plt.ylabel("Marks")

plt.title(
    "Total Average vs Math Average Marks by Gender",
    fontweight='bold'
)

plt.legend()

plt.tight_layout()

plt.show()
```

#### Insights 
- On an average females have a better overall score than men.
- whereas males have scored higher in Maths.

#### BIVARIATE ANALYSIS ( Is Race/Ehnicity has any impact on student's performance ? )

```python
# Group by race/ethnicity and calculate mean scores
Group_data2 = df.groupby('race_ethnicity')[
    ['math_score', 'reading_score', 'writing_score']
].mean().reset_index()

# Create three subplots
fig, ax = plt.subplots(1, 3, figsize=(20, 8))

# Math scores
sns.barplot(
    data=Group_data2,
    x='race_ethnicity',
    y='math_score',
    hue='race_ethnicity',
    palette='mako',
    dodge=False,
    legend=False,
    ax=ax[0]
)

ax[0].set_title('Math score', color='#005ce6', size=20)

for container in ax[0].containers:
    ax[0].bar_label(container, fmt='%.1f', color='black', size=12)


# Reading scores
sns.barplot(
    data=Group_data2,
    x='race_ethnicity',
    y='reading_score',
    hue='race_ethnicity',
    palette='flare',
    dodge=False,
    legend=False,
    ax=ax[1]
)

ax[1].set_title('Reading score', color='#005ce6', size=20)

for container in ax[1].containers:
    ax[1].bar_label(container, fmt='%.1f', color='black', size=12)


# Writing scores
sns.barplot(
    data=Group_data2,
    x='race_ethnicity',
    y='writing_score',
    hue='race_ethnicity',
    palette='coolwarm',
    dodge=False,
    legend=False,
    ax=ax[2]
)

ax[2].set_title('Writing score', color='#005ce6', size=20)

for container in ax[2].containers:
    ax[2].bar_label(container, fmt='%.1f', color='black', size=12)


# Improve chart formatting
for a in ax:
    a.set_ylim(0, 110)
    a.set_xlabel('Race/Ethnicity')
    a.set_ylabel('Average Score')

plt.tight_layout()
plt.show()
```

#### Insights 
- Group E students have scored the highest marks. 
- Group A students have scored the lowest marks. 
- Students from a lower Socioeconomic status have a lower avg in all course subjects

#### 4.4.3 PARENTAL LEVEL OF EDUCATION COLUMN

#### UNIVARIATE ANALYSIS ( What is educational background of student's parent ? )

```python
plt.rcParams['figure.figsize'] = (15, 9)

plt.style.use('fivethirtyeight')

ax = sns.countplot(
    data=df,
    x='parental_level_of_education',
    hue='parental_level_of_education',
    palette='Blues',
    dodge=False
)

# Remove redundant legend
if ax.get_legend() is not None:
    ax.get_legend().remove()

plt.title(
    'Comparison of Parental Education',
    fontweight='bold',
    fontsize=20
)

plt.xlabel('Degree')
plt.ylabel('Count')

# Rotate labels to prevent overlapping
plt.xticks(rotation=30, ha='right')

plt.tight_layout()
plt.show()
```

#### BIVARIATE ANALYSIS ( Is parental education has any impact on student's performance ? )

```python
# Group by parental education and calculate average scores

parental_group = df.groupby('parental_level_of_education')[
    ['math_score', 'reading_score', 'writing_score', 'average']
].mean()

# Plot horizontal bar chart
parental_group.plot(
    kind='barh',
    figsize=(10, 10)
)

plt.title(
    'Average Student Performance by Parental Education',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Average Score')
plt.ylabel('Parental Level of Education')

plt.legend(
    bbox_to_anchor=(1.05, 1),
    loc='upper left',
    borderaxespad=0
)

plt.tight_layout()
plt.show()
```

#### Insights
